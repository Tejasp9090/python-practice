```python
# Given a number n and a digit d, remove one occurrence of
# digit d from n such that the resulting number is maximum.
n = str(input())
d = str(input())
ans = []

for i in range(len(n)):
  if n[i] == d:
    t = n[0:i] + n[(i+1):]
    ans.append(int(t))
print(str(max(ans)))
```

```python
# Given an array of integers, in one operation
# remove either two elements if they are different, or one
# element if they are equal. Find the minimum number of
# operations required to remove all elements.
n = int(input())
arr = list(map(int,input().split()))

ans = 0
arr.sort()

while len(arr) != 0:
  if arr[0] != arr[-1]:
    arr.pop() # last Element
    arr.pop(0) # first element
  else:
    arr.pop(0)
  ans+=1
print(ans)


"""[1,2,3,4,5]
 ↓ remove 1 and 5
[2,3,4]
 ↓ remove 2 and 4
[3]
 ↓ remove 3
[]"""
```
