```python
def fun_1(arr):
  if not arr:
    return 0

  return arr[0]+fun_1(arr[1:])

print(fun_1([1,2,3,4,5,6]))
```

```python
def rotation(s1,s2):
  if len(s1) != len(s2):
    return False
  return s2 in (s1+s2)
print(rotation("Tejas","jasTe"))
```
