```python
# Higher Order Functions
```

```python
# map Operation
salary = [1000,2000,3000,4000,5000]

list(map(lambda sal:sal+sal*.1,salary))
```

```python
salary
```

```python
def test(sal):
  if sal<3000:
    return sal + sal*.1
  elif sal == 3000:
    return sal
  else:
    return sal + sal *.2

list(map(test,salary))
```

```python
# Filter Operation..
tuple(filter(lambda sal:sal>2500,salary))
```

```python
tuple(filter(lambda sal:(sal>2500) or (sal==1000),salary))
```

```python
# reduce Operation
from functools import reduce
```

```python
data=[1,2,3,4,5]
reduce(lambda x,y:x+y,data)
```

```python
list(filter(lambda num:num%2!=0,data))
```
