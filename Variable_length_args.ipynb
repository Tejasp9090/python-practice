```python
# variable length arguments
#1)variable length positional arguments *args
def sample(*num):
  print(num)

sample(10)
sample(10,20,30)
sample()
sample("A")
```

```python
def sample(*num):
  print(num)

sample(10)
sample(10,20,30,"a")
sample()
sample("A")
sample("Tejas")
```

```python
#2) variable length keyword arguments **kwargs

def display(**data):
  print(data)
# in kwargs we must need to pass key=value combination
display(name='Dipali',place = 'Pune',salary=70000)

display(a=100,b=300)
```

```python
# combines *args and **kwrgs
def test(*args,**kwrgs):
  print(args,kwrgs)

test(10,20,30,name="Tejas",place="Karad")
test()
test(13,6,2004,name="Om",Place="Kolhapur")
```

```python
def calculate_salary(basic,bonus):
  total_salary=basic+bonus
  return total_salary


final_salary=calculate_salary(10000,2000)
tax=final_salary*10/100
print("In hand salary:",final_salary-tax)
```

```python
def factorial(n):
  if n ==1 or n==0:
    return 1
  else:
    return n * factorial(n-1)
fact=factorial(5)
fact
```
